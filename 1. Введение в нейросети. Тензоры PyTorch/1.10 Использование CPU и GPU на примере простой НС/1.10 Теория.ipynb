```python
a = torch.tensor([1,2,3])
a.get_device()
# -1 означает ЦПУ, 0,1 и тд означает ГПУ
```

```python
a = torch.tensor([1,2,3], device=device)
a.get_device()
```

```python
torch.cuda.manual_seed(123) # видимо для одного гпу
torch.cuda.manual_seed_all(123) # для всех гпу

# Также для предсказуемости работы нужно ввести следующие команды
torch.backends.cudnn.determenistic = True
torch.backends.cudnn.benchmark = False
```
