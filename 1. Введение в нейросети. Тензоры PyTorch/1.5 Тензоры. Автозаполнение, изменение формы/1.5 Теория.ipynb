Бывает требуется составить тензор из определенных значений

```python
import torch
```

```python
# тензор из нулей
torch.zeros(2, 3, dtype=torch.int32)
```

```python
# тензор из единиц
torch.ones(2, 3, dtype=torch.float64)
```

```python
# единичная матрица
torch.eye(3)
```

```python
# тензор из конкретных значений
torch.full((3, 4), 123)
```

```python
# прогрессия
torch.arange(-3, 7, 2)
```

```python
# разделение интервала на равные отрезки. Указываются границы и число узлов. возвращаются узлы
torch.linspace(1, 5, 4)
```

```python
# равномерное распределение от 0 до 1
torch.rand(2, 3)
```

```python
# стандартизованное нормальное распределение 
torch.randn(2, 3)
# чтобы числа не менялись при каждом перезапуске, можно указать torch.manual_seed(12345)
```

===============================  
в пайторче методы могут иметь нижнее подчеркивание в конце. если имеют, то это inplace методы

```python
t = torch.randn(2, 3)

# заполнение указанным числом
t.fill_(1)
print(t)

# заполнение случайными целыми числами
t.random_(3, 7)
print(t)

# заполнение случайными непрерывными числами
t.uniform_(31, 37)
print(t)

# заполнение случайными непрерывными числами
t.normal_(0, 1)
print(t)
```
