# Подвиг 6.

Используя векторно-матричные операции пакета NumPy, напишите программу, реализующую следующую модель нейронной сети:



Результат вычислений (одно число) выведите в консоль с точностью до десятых.

```python
x1, x2 = list(map(int, input().split())) # чтение вектора [x1, x2] из входного потока
x = np.array([[x1, x2, x1*x2, 1]])

w = np.array([[0.5, 0.5, -1, 2.5]])

y = w @ x.T
print(f"{y.item():.1f}") # вывод выходного значения y с точностью до десятых
```

# Подвиг 8.

Используя векторно-матричные операции пакета NumPy, напишите программу, реализующую следующую модель нейронной сети:



Результат вычислений (одно число
y
y) выведите в консоль с точностью до десятых.

Тесты: https://github.com/selfedu-rus/neuro-pytorch/blob/main/tests/1.2.8

```python
import numpy as np

x1, x2, x3 = list(map(int, input().split())) # чтение вектора [x1, x2, x3] из входного потока

x = np.array([x1, x2, x3, 1])
weights1 = np.array([[0.3, 0.2, 0.5, 0], [0.5, -0.1, -0.2, 2]])

y = x @ weights1.T
y = np.where(y>0, 1, -1)

y = np.append(y, 1)
weights2 = np.array([-1.5, 0.5, -1])
y = y @ weights2

print(f"{y.item():.1f}") # вывод выходного значения y с точностью до десятых
```

# Подвиг 9.

Используя векторно-матричные операции пакета NumPy, напишите программу, реализующую модель нейронной сети в соответствии с формулой:


Результат вычислений (одно число
y
y) выведите в консоль с точностью до десятых.

Тесты: https://github.com/selfedu-rus/neuro-pytorch/blob/main/tests/1.2.9

```python
import numpy as np

x1, x2 = list(map(int, input().split())) # чтение вектора [x1, x2] из входного потока

x = np.array([x1**2, x2**2, 1])
weights = np.array([1, 1, -5])

y = weights @ x

print(f"{y.item():.1f}") # вывод выходного значения y с точностью до десятых
```
