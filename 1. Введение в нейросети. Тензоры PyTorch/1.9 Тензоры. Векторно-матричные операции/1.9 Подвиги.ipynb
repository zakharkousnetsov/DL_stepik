# Подвиг 5. 

На рисунке представлена простейшая модель нейронной сети с одним нейроном и линейной функцией активации на выходе:



  для этой модели:

w = list(map(float, input().split()))  
x = list(map(float, input().split()))  
и формирование на основе этих списков тензоров W, bias и t_inp:

W = torch.tensor([w[1:]], dtype=torch.float32)  
bias = torch.tensor(w[0], dtype=torch.float32)  
t_inp = torch.tensor(x, dtype=torch.float32)  
Тензор W имеет размерность (1, 3) и описывает полносвязный слой из одного нейрона; тензор bias - величина смещения; тензор t_inp имеет размерность (3) и содержит входные значения.

Необходимо продолжить эту программу и реализовать работу этой модели на уровне векторно-матричных операций фреймворка PyTorch согласно следующему выражению:

  

Полученное значение y вывести в консоль в виде вещественного числа с точностью до десятых.

Тесты: https://github.com/selfedu-rus/neuro-pytorch/blob/main/tests/1.9.5

```python
# w = list(map(float, input().split()))
# x = list(map(float, input().split()))

w = [1,2,3,4]
x = [4,3,2]

W = torch.tensor([w[1:]], dtype=torch.float32)
bias = torch.tensor(w[0], dtype=torch.float32)
t_inp = torch.tensor(x, dtype=torch.float32)

y = W.mv(t_inp) + bias

print(f'{y.item():.1f}')
```

# Подвиг 6. 

На рисунке представлена простейшая модель нейронной сети с одним нейроном и сигмоидной функцией активации на выходе:



Матрица весовых коэффициентов W и смещение bias задаются командами:

W = torch.rand(3) * 10 - 5  
bias = torch.rand(1) * 100 - 50  
На вход этой модели подаются сразу несколько векторов, объединенные в пакет (мини-батч) в виде следующей матрицы:

batch_size = 8 # размер мини-батча
X = torch.empty(batch_size, 3).normal_(mean=1.0, std=4.0)  
Необходимо пропустить каждый образ (каждую строку) тензора X через указанную модель нейронной сети и на выходе получить тензор predict размером batch_size с соответствующими значениями.

Подумайте, как это можно эффективно сделать без использования циклов и генераторов.

P.S. На экран ничего выводить не нужно.

```python
W = torch.rand(3) * 10 - 5
bias = torch.rand(1) * 100 - 50

batch_size = 8 # размер мини-батча
X = torch.empty(batch_size, 3).normal_(mean=1.0, std=4.0)

predict = torch.sigmoid(X@W + bias)
```
