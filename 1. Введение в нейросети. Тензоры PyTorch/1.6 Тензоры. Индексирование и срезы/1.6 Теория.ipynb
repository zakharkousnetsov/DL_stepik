```python
import torch
```

```python
a = torch.arange(12)
a
```

```python
a[3]
```

```python
# само значение
a[3].item()
```

```python
# допускаются отрицаиельные индексы
a[-1]
```

```python
# присвоение нового значения
a[0] = 999
a
```

```python
# срезы
b = a[2: 9]
b
```

```python
# срез не создает новый тензор
b[0] = 333
b[-6: -1: 2] = 543
b
```

Обращение к многомерным тензорам

```python
x = torch.arange(9).view(3, -1)
x
```

```python
x[1, -1]
```

```python
x[-1, :]
```

С многомерными тензорами аналогично

```python
x = torch.arange(81).view(3, 3, 3, -1)
x
```

```python
x[0, :, :, :]
```

```python
# центральные столбцы каждой матрицы
x[:, :, :, 1]
```

```python
# аналогично
x[..., 1]
```
