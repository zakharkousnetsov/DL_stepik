# Подвиг 3





```python
def act_sigma(x):
    """
    Реализация функции активации для нейронов скрытого слоя.
    x - тензор [x1, x2]
    Функция должна возвращать тензор типа torch.float32 той же размерности, что и входной тензор x.
    """
    return x**2


def act_u(x):
    """
    Реализация функции активации для нейронов выходного слоя.
    x - тензор [x1]
    Функция должна возвращать тензор типа torch.float32 той же размерности, что и входной тензор x.
    """
    return x


# тензор X в программе не менять
batch_size = 16 # количество входных данных
# X = torch.tensor(list(map(float, input().split())), dtype=torch.float32).view(batch_size, 2)
X = torch.tensor(list(map(float, 2*[1,2,3,4,5,6,7,8,8,7,6,5,4,3,2,1])), dtype=torch.float32).view(batch_size, 2)

W1 = torch.tensor([(1, 0), (1, 0.5)], dtype=torch.float32)
bias1 = torch.tensor([0, 0], dtype=torch.float32)
W2 = torch.tensor([3, -2], dtype=torch.float32)
bias2 = torch.tensor([7.8], dtype=torch.float32)

h = act_sigma(X@W1.transpose(1, 0)+bias1)
predict = act_u(h@W2+bias2)
```

# Подвиг 4





```python
import torch

def act_sigma(x):
    """
    Реализация функции активации для нейронов скрытого слоя.
    x - тензор [x1, x2]
    Функция должна возвращать тензор типа torch.float32 той же размерности, что и входной тензор x.
    """
    return torch.sin(x)


def act_u(x):
    """
    Реализация функции активации для нейронов выходного слоя.
    x - тензор [x1]
    Функция должна возвращать тензор типа torch.float32 той же размерности, что и входной тензор x.
    """
    return torch.where(x>=0.5, torch.tensor(1.0), torch.tensor(-1.0))


# тензор X в программе не менять
batch_size = 16 # количество входных данных
# X = torch.tensor(list(map(float, input().split())), dtype=torch.float32).view(batch_size, 2)
X = torch.tensor(list(map(float, 2*[1,2,3,4,5,6,7,8,8,7,6,5,4,3,2,1])), dtype=torch.float32).view(batch_size, 2)

W1 = torch.tensor([(0.5*torch.pi, 0.5*torch.pi), (0, 0.1)], dtype=torch.float32)
bias1 = torch.tensor([1.5, -1.5*torch.pi], dtype=torch.float32)
W2 = torch.tensor([2, -3.5], dtype=torch.float32)
bias2 = torch.tensor([0], dtype=torch.float32)

h = act_sigma(X@W1.transpose(1, 0)+bias1)
predict = act_u(h@W2+bias2)
predict
```

# Подвиг 5





```python
import torch

def sigma(x):
    """
    Реализация функции активации для нейронов выходного слоя.
    x - тензор [x1]
    Функция должна возвращать тензор типа torch.float32 той же размерности, что и входной тензор x.
    """
    return torch.where(x>0.0, torch.tensor(1.0), torch.tensor(-1.0))

# тензор X в программе не менять
batch_size = 32 # количество входных данных
# X = torch.tensor(list(map(float, input().split())), dtype=torch.float32).view(batch_size, 2)
X = torch.tensor(list(map(float, 4*[1,2,3,4,5,6,7,8,8,7,6,5,4,3,2,1])), dtype=torch.float32).view(batch_size, 2)

W1 = torch.tensor([(3, -4), (9, -5)], dtype=torch.float32)
bias1 = torch.tensor([18, 12], dtype=torch.float32)
W2 = torch.tensor([-1, -1], dtype=torch.float32)
bias2 = torch.tensor([1], dtype=torch.float32)

predict = sigma((sigma(X@W1.transpose(1, 0)+bias1))@W2+bias2)
predict

```
