```python
import torch
```

```python
a = torch.tensor([1,2,3])
a - 3
```

```python
-a
```

Можно выполнять операцию между тензорами

```python
a - 2*a
```

Тензоры должны быть согласованы по длине

```python
b = torch.tensor([1,2,3, 4])
```

```python
a+b
```

Но возможна такая ситуация. Тогда операция пройдет по оси - это транслирование тензоров

```python
a = torch.tensor([10, 10, 10])
b = torch.arange(6).view(2, -1)

a+b
```

Как я понял, важно, чтобы у слагаемых тензоров совпадали последние размерности

```python
a = torch.arange(1, 19).view(3, 3, 2)
b = torch.ones(3, 2)
```

```python
a+b
```
