```python
class TwoLayerModel(nn.Module):
    def __init__(self, inp=3, hidd=2, outp=1):
        super().__init__()
        self.layer1 = nn.Linear(inp, hidd)
        self.f_act = F.sigmoid
        self.layer2 = nn.Linear(hidd, outp)

    def forward(self, x):
        x = self.layer1(x)
        x = self.f_act(x)
        x = self.layer2(x)
        return x

x = torch.rand(3)

model = TwoLayerModel()
model.eval()
predict = model(x)
predict
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class ThreeLayersModel(nn.Module):
    def __init__(self, n1, n2, n3, n4):
        super().__init__()
        self.layer1 = nn.Linear(n1, n2)
        self.layer2 = nn.Linear(n2, n3)
        self.layer3 = nn.Linear(n3, n4)
        self.f_act = F.tanh

    def forward(self, x):
        x = self.layer1(x)
        x = self.f_act(x)
        x = self.layer2(x)
        x = self.f_act(x)
        x = self.layer3(x)
        return x

model = ThreeLayersModel(32, 10, 12, 1)
batch_size = 4
x = torch.rand(batch_size, 32)

model.eval()

predict = model(x)
predict
```
