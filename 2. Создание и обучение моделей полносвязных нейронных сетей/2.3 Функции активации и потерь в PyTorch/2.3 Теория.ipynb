# Функции активации

Самые популярные ФА доступны в модуле torch.nn.functional

```python
import torch.nn.functional as F
```

самыми первыми стали использовать сигмоиду и гиперболический тангенс. По сути они отличаются только диапазоном выхода. Сейчас их применяют либо в маленьких сетях, либо только на выходных слоях. при больших значениях аргумента в этих функцияъ практически отсутствует уклон, поэтому градиент оказывает очень мал, и обучение не происходит



Поэтому чаще используют следующие функции активации (особенно чисто RelU). Они все не имеют пологих участков в положительных областях



На выходном слое для задачи регресси используют F.Linear (просто полносвязный слой, тк он линейный, его глупо использовать для промежуточных слоев, иначе все слои просто схлопнутся)   

для классификации F.sigmoid (бинарная) / F.sofmax(многоклассовая)



# Функции потерь

==Регрессия==  
nn.MSELoss - mse  
nn.L1Loss - mae  

==Классификация==  
nn.BCELoss - бинарная кросс-энтропия (для задач бинарной классификации)  
nn.BCEWithLogitsLoss - бинарная кросс-энтропия после применения сигмоидной функции потерь. Короче на выходе можно не писать отдельную сигмоидную ФА. Но преимущество еще и в том, что это более усточивое решение, чем отдельно BCELoss+sigmoid. Если используется эта ФП, то на выходном нейроне нужно применять линейную функции активации, то есть не искажать сигнал (а зачем она тогда вооббще нужна)  
nn.CrossEntropyLoss - для задач многоклассовой классификации
nn.NLLLoss - тоже для задач многоклассовой классификации, но используется реже
