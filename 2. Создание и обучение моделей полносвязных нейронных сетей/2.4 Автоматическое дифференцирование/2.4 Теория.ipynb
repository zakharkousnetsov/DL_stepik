```python
import torch
```

В pytorch дифференцирование реализовано численно  
https://en.wikipedia.org/wiki/Automatic_differentiation

```python
x = torch.tensor([2.0], requires_grad=True)
y = torch.tensor([-4.0], requires_grad=True)

f = (x+y)**2 + 2*x*y
f.backward()

print(f'значение самой функции {f}')
print(f'значение переменной x {x.data}, значение производной по x в указанной точке {x.grad}')
print(f'значение переменной y {y.data}, значение производной по y в указанной точке {y.grad}')
```
