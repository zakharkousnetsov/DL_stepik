Для того, чтобы минимизировать ошибку между предсказаниями конкретного класса, автоенкодер по суи усредняет все изображения класса, поэтому изображение получается смазанным  

  
Делаем вывод, что критерий MSE не совсем подходит для оценки качества реконтрукции

Идея в том, чтобы заставить еще одну сеть угадывать, является ли поданное ей на вход изображение реальным или сгенерированным. 
  
На вход дискриминатора будем подавать батч реальных изображений, а потом батч сгенерированных. Реальные изображения будут отмечены как 1, а сгенерированные как 0  

Выход дискриминатора, когда подаются реальные изображения, будем называть real_out, а когда сгенерированные - fakeout. Дискриминатор будет обучаться совместно с генератором. Качество предсказания дискриминатора будет считаться через бинарную кросс-энтропию. Для каждого батча изображений мы будем подавать реальное изображение и формировать real_out, а потом подавать сгенерированное изображение и формировать fake_out. Целевое значение дискриминатора, когда подаются реальные изображения, обозначим tr=1, когда сгенерированные - tf=0. Итоговая общая функция потерь выглядит следующим образом:



как обучать генератор? генератор должен генерировать такие изображения, чтобы дискриминатор посчитал их реальными, то есть на выходе дискриминатора было 1. То есть мы прогоняем входной сигнал через генератор и дискриминатор, получаем выход и в соответствии с ним корректируем веса генератора. Функция потерь как у дискриминатор, только с tf=1
