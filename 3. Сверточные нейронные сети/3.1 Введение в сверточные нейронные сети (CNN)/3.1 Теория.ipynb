Цветные картинки обрабатываются трехмерным ядром. в результате каждое ядро умножается на свой канал и результаты складываютс как бы в одну карту активации. к ней прибавляет биас и применяется функция активации

Пулинг служит для снижения размерности карта активаций, а главное при этом выделенные значения (максимумы) как бы сближаются и могут на следующем шаге быть обработаны одним ядром, то есть признаки могут быть обобщены
