В прошлый раз мы кодировали буквы через ono-hot вектора. Но если так кодировать слова, то матрица получится слишком разряженной. А также между словами не будет никакой связи. так как расстояние между всеми точками пространства всегда будет = 1. Вместо этого мы могли бы ввести более низкоразмерное пространство, и тогда внутри него станет можно размещать новые слова, даже не участвовашие в его построении

эмбеддинг - векторное представление объекта в некотором пространстве, отражающий его положение относительно других объектов. Перевод слова в векторное пространство называется word2vec. Подходов много

skip-gram - предсказание контекста по центральному слову  
CBOW - предсказание центрального слова по контексту    



one-hot представление слова умножается на матрицу и получается более низкоразмерное представление

чем нам может помочь приведение к эмбеддингу в обучении нейронных сетей

Умножение на матрицу W здесь можно считать эмбеддингом, то есть приведением к более низкоразмерному пространству. Однако когда словарь большой, то лучше отдельно выделить слой эмбеддинга. То есть по сути у нас ставится два линейных слоя подряд, что ничем не лучше одного линейного слоя. Но если слой эмбеддинга уже предобучен выделять ключевой смысл в объектах, то это поможет
