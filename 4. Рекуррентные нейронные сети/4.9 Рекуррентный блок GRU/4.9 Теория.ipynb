LSTM хороша, но имеет много обучаемых параметров. В качестве альтернативы предложена модель GRU

Здесь всего один вектор скрытого состояния. А чтобы определить, что забыть, а что оставить, используется вектор zt. Просто в случае с забыванием он используется в виде (1-zt)



В LSTM и GRU меньше проявляется затухание градиента. Но зато у них вше риск переобучения

```python
import torch
import torch.nn as nn
 
rnn = nn.GRU(10, 20, batch_first=True)
 
x = torch.randn(7, 3, 10)    # (batch_size, sq_length, d_size)
y, h = rnn(x)
 
print('y:', y.size())
print('h:', h.size())
```
