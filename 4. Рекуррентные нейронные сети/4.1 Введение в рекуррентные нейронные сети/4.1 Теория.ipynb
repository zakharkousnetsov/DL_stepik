Способ обработки последовательности: подавать в некоторую функцию текущее значение и выход этой же функции с предыдущего шага

В данном случае подается вектор из нескольких последовательных значений (скользящее окно), которое складывается с вектором h предыдущего шага, а также отправляется в решающую голову
